```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import sklearn 
```

```python
rides = pd.DataFrame({
    "distance": [2, 5, 3, 8, 12, 6, 15, 4, 10, 7],
    "price":    [12, 21, 14, 30, 41, 24, 52, 18, 37, 27],
})
rides.index = rides.index + 1  
rides
```

```python
plt.figure(figsize=(10,7))
plt.scatter(rides["distance"],rides["price"],color="green",s=70)
plt.xlabel("distance(KM)")
plt.ylabel("price(MAD)")
plt.title("Price/Distance")
plt.grid(alpha=0.3)
plt.show()
```

```python
def predict(x,m,p):
    return m*x+p
def sse(x,y,m,p):
    gaps=y-predict(x,m,p)
    return np.sum(gaps**2)
def show_line(x,y,m,p):
    plt.figure(figsize=(12,7))
    plt.scatter(x,y,color="green",s=70,label="Real value")
    plt.scatter(x,predict(x,m,p),color="red",s=70,label="Our Predictions")
    for x_true,y_true,y_pred in zip(x,y,predict(x,m,p)):
        plt.plot([x_true,x_true],[y_true,y_pred],color='gray',linestyle='--',alpha=0.6)
    xs=np.array([x.min()-1,x.max()+1])
    plt.plot(xs,predict(xs,m,p),color='red',alpha=0.5)
    plt.xlabel('Distance(KM)')
    plt.ylabel('Price(MAD)')
    plt.legend()

```

```python
x=rides["distance"]
y=rides["price"]
```

```python
m=3
p=2.75
show_line(x,y,m,p)
```
